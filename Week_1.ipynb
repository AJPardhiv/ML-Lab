**Post-Lab Programs**

```python
# Post-Lab 1: Sum of Digits of a Given Number
num = int(input("Enter a number: "))
sum_digits = 0

while num > 0:
    digit = num % 10
    sum_digits += digit
    num //= 10

print("Sum of digits:", sum_digits)
```

```python
# Post-Lab 2: Multiplication Table up to 10
num = int(input("Enter a number: "))

print(f"Multiplication Table of {num}:")
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
# Post-Lab 3: Pie Chart for Categorical Data Distribution (Survived Column)
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/content/Titanic-Dataset (1).csv")

# Count categorical values (Survived: 0 = No, 1 = Yes)
survival_counts = data["Survived"].value_counts()

plt.pie(survival_counts, labels=["Not Survived", "Survived"], autopct="%1.1f%%")
plt.title("Survival Distribution in Titanic Dataset")

plt.show()
```

```python
# Post-Lab 4: Compare Performance of Different Data Loading Methods in pandas
import pandas as pd
import time

file_path = "/content/Titanic-Dataset (1).csv"

# Method 1: read_csv
start_time = time.time()
data1 = pd.read_csv(file_path)
end_time = time.time()
print("Time taken by read_csv:", end_time - start_time, "seconds")

# Method 2: read_table
start_time = time.time()
data2 = pd.read_table(file_path, delimiter=",")
end_time = time.time()
print("Time taken by read_table:", end_time - start_time, "seconds")
```

```python
# Post-Lab 5: Error Handling for File Operations
import pandas as pd

file_path = "/content/Titanic-Dataset (1).csv"

try:
    data = pd.read_csv(file_path)
    print("File loaded successfully!")
    print(data.head())

except FileNotFoundError:
    print("Error: File not found. Please check the file path.")

except pd.errors.EmptyDataError:
    print("Error: The file is empty.")

except Exception as e:
    print("An unexpected error occurred:", e)
```
